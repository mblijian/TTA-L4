```python
# face recognition
def face_diagnoal():
    import time
    got.load_models(["face_recognition"])
    time.sleep(1)
    while True:
        face_info = got.get_face_recognition_total_info()
        if face_info:
            face_x = face_info[0][1]
            face_width = face_info[0][4]
            if face_width>70 and 280<face_x<380:
                got.mecanum_stop()
                break
            elif face_x > 380: # face too much right
                got.mecanum_move_xyz(15,5,0)
            elif face_x < 280: # face too much left
                got.mecanum_move_xyz(-15,5,0)
            else: # face in the middle
                got.mecanum_move_xyz(0,5,0)
        else:
            got.mecanum_move_xyz(0, 10, 0)

# line follow
def line_follow():
    import time
    got.load_models(["line_recognition"])
    got.set_track_recognition_line(0)

    while True:
        line_info = got.get_single_track_total_info()
        offset = line_info[0]
        line_type = line_info[1]

        if line_type> 0:
            got.mecanum_move_xyz(0,20,int(0.25*offset))
            time.sleep(0.1)
        else:
            got.mecanum_stop()
            break
    print("There's no line to follow,let's stop!")

# pick up function
def pick_up():
    import time
    got.mechanical_clamp_release()
    time.sleep(2)
    got.mechanical_joint_control(0,30,30, 1000)
    time.sleep(1)
    got.mechanical_joint_control(0,0,-70, 1000)
    time.sleep(1)
    got.mechanical_clamp_close()
    time.sleep(2)
    got.mechanical_joint_control(90,30,30, 1000)
    time.sleep(1)

# put down function
def put_down():
    import time
    got.mechanical_joint_control(0,30,30, 1000)
    time.sleep(1)
    got.mechanical_joint_control(0,0,-80, 1000)
    time.sleep(1)
    got.mechanical_clamp_release()
    time.sleep(2)
    got.mechanical_joint_control(0,30,30, 1000)
    time.sleep(1)
    got.mechanical_clamp_close()
    time.sleep(2)

# Apriltag diagnoal
def AP_diagnoal():
    import time
    got.load_models(["apriltag_qrcode"])
    time.sleep(1)
    while True:
        AP_info = got.get_apriltag_total_info()
        if AP_info:
            AP_x = AP_info[0][1]
            AP_distance = AP_info[0][6]

            if AP_distance<0.2 and 280<AP_x<380:
                got.mecanum_stop()
                break
            elif AP_x > 380: # face too much right
                got.mecanum_move_xyz(10,5,0)
            elif AP_x < 280: # face too much left
                got.mecanum_move_xyz(-10,5,0)
            else: # face in the middle
                got.mecanum_move_xyz(0,5,0)
        else:
            got.mecanum_move_xyz(0, 10, 0)
```

```python
import time
AP_diagnoal()
time.sleep(1)
pick_up()
time.sleep(1)
line_follow()
time.sleep(1)
face_diagnoal()
time.sleep(1)
put_down()
```
